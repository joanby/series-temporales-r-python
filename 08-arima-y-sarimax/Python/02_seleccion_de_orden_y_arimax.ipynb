```python
import pandas as pd

ftse = pd.read_csv("ftse100_close_2015_2026.csv", parse_dates=["date"], index_col="date")["close"]
spx = pd.read_csv("sp500_close_2015_2026.csv", parse_dates=["date"], index_col="date")["close"]

print("FTSE:", ftse.shape, ftse.index.min(), ftse.index.max())
print("S&P 500:", spx.shape, spx.index.min(), spx.index.max())
```

```python
from statsmodels.tsa.arima.model import ARIMA

resultados = []
for p in range(1, 6):
    modelo = ARIMA(ftse, order=(p, 1, 1))
    ajuste = modelo.fit()
    resultados.append({"order": f"({p},1,1)", "log_likelihood": ajuste.llf, "aic": ajuste.aic, "bic": ajuste.bic})

tabla = pd.DataFrame(resultados)
tabla
```

```python
mejor = tabla.loc[tabla["aic"].idxmin()]
print("mejor orden por AIC:", mejor["order"])
print(tabla.to_string(index=False))
```

```python
modelo_411 = ARIMA(ftse, order=(4, 1, 1))
resultado_411 = modelo_411.fit()
print(resultado_411.summary())
```

```python
spx_ret = spx.pct_change().dropna() * 100

datos = pd.concat([ftse, spx_ret], axis=1, join="inner")
datos.columns = ["ftse_close", "spx_ret"]
print(datos.shape)
datos.head()
```

```python
modelo_arimax = ARIMA(datos["ftse_close"], order=(1, 1, 1), exog=datos["spx_ret"])
resultado_arimax = modelo_arimax.fit()
print(resultado_arimax.summary())
```

```python
print("Comparacion de AIC:")
print("  ARIMA(1,1,1) sin exogena :", 33207.72)
print("  ARIMA(4,1,1) sin exogena :", round(resultado_411.aic, 2))
print("  ARIMAX(1,1,1) con S&P500 :", round(resultado_arimax.aic, 2))
```
